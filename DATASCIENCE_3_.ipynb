# ***pandas***

```python
import pandas as pd
data=[20,40,60,80,100]
s=pd.Series(data,index=['A','B','C','D','E'])
print(s)
```

```python
data1={'name':'anandhu','age':19}
s=pd.Series(data1)
print(s)
```

```python
data2=[12,23,34,45,56]                    # loc
s=pd.Series(data2)
print(s.loc[1])
```

```python
data2=[12,23,34,45,56]                              # iloc
s=pd.Series(data2,index=['a','b','c','d','e'])
print(s.iloc[1])
```

```python
data3=[10,20,30,40]
s=pd.Series(data3,index=["a","b","c","d"])         # slicing using
print(s.loc["a":"b"])                             # loc
print(s.iloc[:2])                                # iloc
```

```python
data3=[10,20,30,40]
s=pd.Series(data3,index=["a","b","c","d"])      # slicing using
print(s.loc["a":"b"])                             # loc
print(s.iloc[:2])                                # iloc
print(s.loc[s>10])
```

```python
import pandas as pd
sales=pd.Series(
    [1200,850,1500,950,2000],
    index=["Monday","Tuesday","Wednesday","Thursday","Friday"]
)
print(sales)
print(sales.max())
print(sales.sum())
print(sales[sales<=1000])
```

```python
data=[20,20,30,30,50,60,60,80]
arr=pd.Series(data)
print(arr)
print(arr.unique())                 #remove repeated no.s
print(arr.nunique())                #count
print(arr.value_counts())           #reapted no.s count

```

```python
data=[None,20,30,None,50,None,60,80]
arr=pd.Series(data)
print(arr)
print(arr.isna())
print(arr.notnull())
print(arr.dropna())
print(arr.isnull())
print(arr.notna())
```
